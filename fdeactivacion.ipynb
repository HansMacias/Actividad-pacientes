```python
import pandas as pd

# URL del archivo CSV en GitHub
csv_url = 'https://raw.githubusercontent.com/Fernoguez/EEG/main/DM.csv'

# Cargar el CSV en un DataFrame de pandas
df = pd.read_csv(csv_url)

# Mostrar las primeras 5 filas del DataFrame
display(df.head())
```

A continuación, se definen las 5 funciones de activación principales mencionadas en el documento:

1.  **ReLU** (Rectified Linear Unit)
2.  **Sigmoid** (Función Logística)
3.  **Softmax** (Función Softmax)
4.  **Tanh** (Tangente Hiperbólica)
5.  **Leaky ReLU** (Leaky Rectified Linear Unit)

```python
import numpy as np

# 1. ReLU (Rectified Linear Unit)
def relu(x):
    return np.maximum(0, x)

# 2. Sigmoid (Función Logística)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 3. Softmax (Función Softmax)
# Softmax se aplica típicamente a un vector de entradas, no a un solo escalar.
# Para aplicarla a una serie de Pandas, la trataremos como un vector para cada fila,
# aunque la naturaleza de las columnas de EEG (como 'Volts') sugiere que cada
# entrada es independiente. Para este ejercicio, la aplicaremos como si cada
# valor fuera un 'logit' de una sola clase, aunque su uso principal es para
# clasificaciones multiclase donde la suma de las salidas es 1.
# Si quisieras aplicar Softmax correctamente a múltiples características,
# necesitaríamos un vector de logits por fila.
def softmax_single_value(x):
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x) # Esto solo tiene sentido si 'x' es un array

# Vamos a definirla para que sea aplicable a una serie como si cada elemento
# fuera un logit en un contexto de softmax donde solo él es el 'ganador' potencial.
# Para la aplicación a una columna individual, la interpretación será que
# normaliza los valores en relación a sí mismos, lo cual no es el uso típico de softmax.
# Una implementación más útil para una única columna sería normalizar los valores
# de la columna entre 0 y 1, pero eso no es Softmax.
# Por simplicidad y para mostrar una "aplicación" directa,
# simplemente usaremos la exponencial normalizada por la suma de TODA la serie.
# Esto NO es el uso canónico de Softmax para clasificación multi-clase.
# Para este ejercicio, lo aplicaremos a una serie, y cada valor se normalizará respecto a la suma de exponenciales de la serie completa.
# Esto es una simplificación y no representa el uso típico de Softmax en capas de salida.
# Una mejor alternativa sería aplicar Softmax a un conjunto de columnas que representen logits.
# Dado que solo tenemos 'Volts', la interpretación será algo forzada aquí.
# Dejaremos la implementación tal cual, y se notará que no es el uso habitual de Softmax.

# 4. Tanh (Tangente Hiperbólica)
def tanh(x):
    return np.tanh(x)

# 5. Leaky ReLU (Leaky Rectified Linear Unit)
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, x * alpha)
```

Ahora aplicaremos estas funciones a la columna 'Volts' del DataFrame `df` y crearemos nuevas columnas con los resultados.

```python
# Asumiendo que 'C3' es la columna numérica a la que queremos aplicar las funciones de activación
# Si hay otra columna que te interese, por favor indícala.
input_column = 'C3'

# Aplicar ReLU
df['C3_ReLU'] = df[input_column].apply(relu)

# Aplicar Sigmoid
df['C3_Sigmoid'] = df[input_column].apply(sigmoid)

# Aplicar Softmax
# Advertencia: La aplicación de softmax a una única columna como esta no es su uso típico.
# Softmax normalmente se aplica a un vector de logits (salidas no normalizadas)
# para obtener una distribución de probabilidad sobre múltiples clases.
# Aquí, la aplicaremos de una forma que cada valor se normaliza respecto a la suma de exponenciales de toda la columna.
# Esto es para demostrar la función, pero no su aplicación común en redes neuronales.
exp_values = np.exp(df[input_column])
sum_exp_values = np.sum(exp_values)
df['C3_Softmax'] = exp_values / sum_exp_values

# Aplicar Tanh
df['C3_Tanh'] = df[input_column].apply(tanh)

# Aplicar Leaky ReLU
df['C3_LeakyReLU'] = df[input_column].apply(leaky_relu)

# Mostrar las primeras filas del DataFrame con las nuevas columnas
display(df.head())
```

La columna 'C3' contiene valores que oscilan aproximadamente entre 1.3 y 1.5, todos ellos positivos. Esta característica es crucial para entender el comportamiento de algunas de las funciones de activación.

## Análisis por Función de Activación

### 1. ReLU (Rectified Linear Unit)

**Definición:** `f(x) = max(0, x)`

**Resultados en 'C3':** Al aplicar ReLU a la columna 'C3', se observó que los valores resultantes eran idénticos a los originales. Esto se debe a que todos los valores en 'C3' son positivos (mayores que cero). Por lo tanto, para cada `x` en 'C3', `max(0, x)` simplemente devuelve `x`.

**Implicaciones:** En este contexto particular, donde todas las entradas son positivas, ReLU actúa como una función de identidad, no introduciendo cambios en los valores. Si la columna 'C3' hubiera contenido valores negativos, estos se habrían convertido en cero.

### 2. Sigmoid (Función Logística)

**Definición:** `f(x) = 1 / (1 + e^(-x))`

**Resultados en 'C3':** Los valores de la columna 'C3' (aproximadamente entre 1.3 y 1.5) fueron transformados por la función Sigmoid a un rango muy estrecho de valores entre 0 y 1. Específicamente, se obtuvieron valores cercanos a 0.8. Esto indica que los valores originales se encuentran en la región de saturación positiva de la función sigmoide.

**Implicaciones:** Sigmoid comprime los valores en el rango (0, 1), lo que la hace útil para la salida de probabilidades en clasificación binaria. Sin embargo, para valores de entrada relativamente grandes (tanto positivos como negativos), la función se 'satura', lo que significa que el gradiente se vuelve muy pequeño. Esto puede llevar al problema de desvanecimiento del gradiente en el entrenamiento de redes neuronales profundas.
